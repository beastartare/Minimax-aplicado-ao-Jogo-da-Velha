# Como funciona o algoritmo Minimax

O Minimax é um algoritmo utilizado para tomar decisões em jogos entre dois jogadores. Ele funciona analisando as possíveis jogadas e seus respectivos resultados futuros, considerando que os dois jogadores sempre tentarão fazer a melhor escolha possível.

Para isso, o algoritmo representa cada situação do jogo como um estado e simula as jogadas seguintes, formando uma espécie de árvore de possibilidades. Em cada nível dessa árvore, um dos jogadores escolhe sua jogada.

No jogo da velha, por exemplo, podemos considerar X como MAX e O como MIN. O jogador MAX procura obter o maior valor possível, enquanto o jogador MIN procura obter o menor valor. Os resultados podem ser representados por 1 para uma vitória de X, 0 para um empate e -1 para uma vitória de O.

O algoritmo continua analisando as possibilidades até chegar a um estado em que o jogo terminou. A partir desses resultados, os valores são retornados pela árvore: MAX escolhe o maior valor entre suas possibilidades, enquanto MIN escolhe o menor. Dessa forma, cada jogador considera não apenas sua própria jogada, mas também a melhor resposta que o adversário poderia dar.

Ao final da análise, o Minimax consegue determinar qual é a melhor jogada para o jogador atual, considerando todas as possibilidades futuras do jogo.

# Algoritmo

```python

def pontuação(tabuleiro):
  vencedor = verificar_vencedor(tabuleiro)
  if vencedor == "X": #max
    return 1
  elif vencedor == "O": #min
    return -1
  else:
    return 0

def minimax(tabuleiro, jogador):

  #verifica se o jogo terminou (parada da recursão) e retorna o vencedor
  if jogo_terminou(tabuleiro):
    return pontuação(tabuleiro), None

  if jogador == "X":
    #começa-se considerando o pior valor possivel para o max(X). Nesse caso, -infinito
    melhor_valor = -float("inf")
    melhor_jogada = None

    for jogada in jogadas_disponiveis(tabuleiro):
      # a copia do tabuleiro é realizada para fazer o teste das possiveis jogadas sem alterar o tabuleiro original
      novo_tabuleiro = tabuleiro.copy()
      #testa cada jogada
      novo_tabuleiro[jogada] = jogador

      # função recursiva para verificar o que o jogador "O" fará para assim decidir a melhor jogada
      valor, _ = minimax(novo_tabuleiro, "O")

      # escolherá o melhor valor (nesse caso, o maior)
      if valor > melhor_valor:
        melhor_valor = valor
        melhor_jogada = jogada

    return melhor_valor, melhor_jogada

    # Mesma logica para o jogador "O"
  else:

    melhor_valor = float("inf")
    melhor_jogada = None

    for jogada in jogadas_disponiveis(tabuleiro):

        novo_tabuleiro = tabuleiro.copy()

        novo_tabuleiro[jogada] = jogador

        valor, _ = minimax(novo_tabuleiro, "X")

        if valor < melhor_valor:
          melhor_valor = valor
          melhor_jogada = jogada

    return melhor_valor, melhor_jogada

  return None


```

# Jogo da Velha

```python
#lista que armazena as possibilidades de vitorias (diagonais, horizontal, vertical)
vitorias = [
    [0, 1, 2],
    [3, 4, 5],
    [6, 7, 8],
    [0, 3, 6],
    [1, 4, 7],
    [2, 5, 8],
    [0, 4, 8],
    [2, 4, 6]
]

#função que verifica quem venceu (X ou O)
def verificar_vencedor(tabuleiro):
    for combinacoes in vitorias:
        if tabuleiro[combinacoes[0]] == tabuleiro[combinacoes[1]] == tabuleiro[combinacoes[2]] != " ":
            return tabuleiro[combinacoes[0]]

    return None

#verifica quais jogadas ainda estão disponiveis
def jogadas_disponiveis(tabuleiro):
  jogadas  = []

  for i in range(9):
    if tabuleiro[i] == " ":
      jogadas.append(i)
  return jogadas

#verificar empate
def empate(tabuleiro):

    if not jogadas_disponiveis(tabuleiro) and verificar_vencedor(tabuleiro) == None:
        return True
    return False

#verifica termino do jogo
def jogo_terminou(tabuleiro):
  if verificar_vencedor(tabuleiro) is not None or len(jogadas_disponiveis(tabuleiro))==0:
    return True

  return False

#função para mostrar o tabuleiro
def mostrar_tabuleiro(tabuleiro):
    print()
    print(tabuleiro[0], "|", tabuleiro[1], "|", tabuleiro[2])
    print("---------")
    print(tabuleiro[3], "|", tabuleiro[4], "|", tabuleiro[5])
    print("---------")
    print(tabuleiro[6], "|", tabuleiro[7], "|", tabuleiro[8])

def jogar():

  #lista que armazena as posições do jogo da velha (9 posições)
  tabuleiro = [" ", " ", " ",
             " ", " ", " ",
             " ", " ", " "]

  jogador = "O"
  #para testar um estado inicial diferente
  '''tabuleiro[4] = "X"
  tabuleiro[5] = "O"
  tabuleiro[2] = "X"'''


  while not jogo_terminou(tabuleiro):
    mostrar_tabuleiro(tabuleiro)
    valor, jogada = minimax(tabuleiro, jogador)
    tabuleiro[jogada] = jogador

    if jogador == "X":
      jogador = "O"
    else:
      jogador = "X"

  mostrar_tabuleiro(tabuleiro)


  if verificar_vencedor(tabuleiro)=="X":
    print("\n O jogador_X venceu")
  elif verificar_vencedor(tabuleiro)=="O":
    print("\n O jogador_O venceu")
  else:
    print("Empate")

jogar()


```
